```python
import pandas as pd
import numpy as np
import hashlib

# Going up one directory to find the CSV
df = pd.read_csv(r"C:\Users\Noman Traders\Downloads\teaching_prices_clean (1).csv")
print("Successfully loaded! Shape:", df.shape)
```

```python
rows = []
for item, g in df.groupby("item", sort=False):
    g = g.sort_values(["date", "city"])
    p_first = g[g["date"] == g["date"].min()]["price"].mean()
    p_last = g[g["date"] == g["date"].max()]["price"].mean()
    pct_change = ((p_last - p_first) / p_first) * 100
    rows.append({
        "item": item,
        "mean": round(g["price"].mean(), 2),
        "median": round(g["price"].median(), 2),
        "opening_avg": round(p_first, 2),
        "closing_avg": round(p_last, 2),
        "pct_change": round(pct_change, 2)
    })
```

```python
summary_df = pd.DataFrame(rows)
print("\n=== SUMMARY STATISTICS ===")
print(summary_df.to_string(index=False))

# Week 2 flagged-value sensitivity check on Staple A
staple_clean = df[df["item"] == "Staple A"].sort_values(["date", "city"])["price"].tolist()
staple_flagged = staple_clean.copy()
staple_flagged[-2] = 1145.0  # Decimal-shift typo test: 114.5 -> 1145.0

print("\n=== WEEK 2 FLAGGED CHECK (STAPLE A) ===")
print(f"With flagged value    -> Mean: {np.mean(staple_flagged):.2f}, Median: {np.median(staple_flagged):.2f}")
print(f"Without flagged value -> Mean: {np.mean(staple_clean):.2f}, Median: {np.median(staple_clean):.2f}")

# Generate SHA-256 commit hash
draft_text = "Over the twelve weekly observations from 1 January to 19 March 2026 across City 1 and City 2, prices rose for all four tracked items in the teaching dataset."
commit_hash = hashlib.sha256(draft_text.encode("utf-8")).hexdigest()[:12]
print("\n=== COMMIT HASH ===")
print(f"SHA-256 Hash: {commit_hash}")
```
